## Задание

Необходимо вычислить значение выражения:
$$Z=\frac{1}{1\cdot3}-\frac{X}{2\cdot4}+\frac{X^2}{3\cdot5}-\frac{X^3}{4\cdot6}+...$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $X$. Переменная, которая должна быть вычислена на основе независимых: $Z$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для представления количества слагаемых.

Обратите внимание, что знаки слагаемых чередуются, а степени числителя и множители знаменателя привязаны к порядковому номеру слагаемого.

## Вычисление выражения, печать результатов

Код для вычисления значения выражения $Z$:

```python
X = 2
N = 10
Z = 0
for i in range(1, N+1):
  chisl = X**(i-1)
  znam = i * (i+2)
  if i%2==1:
    Z += chisl / znam
  else:
    Z -= chisl/znam
print(Z)
```

Решите пример в соответствии с индивидуальным заданием (Вариант 11):
$$p = -\frac{X \cdot \sin(Y)}{1 \cdot 2}+\frac{X^{3} \cdot \cos(Y^{2})}{3 \cdot 4}-\frac{X^{5} \cdot \sin(Y^{3})}{5 \cdot 6}+\frac{X^{7} \cdot \cos(Y^{4})}{7 \cdot 8}-\ldots$$

```python
import math
X = 0.5
Y = 1
N = 10
p = 0
for i in range(1, N+1):
  x_pow = X**(2*i-1)

  if i%2 == 1:
    trig = math.sin(Y**i)
  else:
    trig = math.cos(Y**i)

  denominator = (2*i-1)*(2*i)
  sign = (-1)**i
  term = sign * x_pow * trig / denominator
  p += term
print(p)
print("Type of X:", type(X))
print("Type of Y:", type(Y))
print("Type of N:", type(N))
print("Type of p (end):", type(p))
```
