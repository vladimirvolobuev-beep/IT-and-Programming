## Задание

Напишите функцию для вычисления суммы квадратов натуральных чисел от **a** до **b**, где **a** и **b** – натуральные числа. Продемонстрируйте использование объявленной функции.

## Решение

### Объявление функции

```python
def sum_ab(a, b):
  s = 0
  for i in range(a, b+1):
    s+=i*i
  return s
```

### Демонстрация использования функции

```python
x, y = 1, 2
print(f'Сумма квадратов чисел от {x} до {y}: {sum_ab(x, y)}')
x, y = 500, 50000
print(f'Сумма квадратов чисел от {x} до {y}: {sum_ab(x, y)}')
x, y = 268, 3258
print(f'Сумма квадратов чисел от {x} до {y}: {sum_ab(x, y)}')
```

Решите пример в соответствии с индивидуальным заданием (Вариант 11):

Напишите функцию, которая возвращает **True**, если переданное ей число является квадратом некоторого натурального числа, в противном случае - **False**.

```python
def square(x):
  if x <= 0:
    return False
  root = int(x ** 0.5)
  if root ** 2 == x:
    return True
  else:
    return False

# Демонстрация использования функции
x = 1
print(f"Число {x} является квадратом натурального числа: {square(x)}")
x = 6
print(f"Число {x} является квадратом натурального числа: {square(x)}")
x = -4
print(f"Число {x} является квадратом натурального числа: {square(x)}")
x = 25
print(f"Число {x} является квадратом натурального числа: {square(x)}")
```
