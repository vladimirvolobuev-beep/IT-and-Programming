# **Ответы на контрольные вопросы**

**1.** Переменная в Python - это имя (ссылка), связанное с объектом в памяти.
Присваивание: справа вычисляется значение, затем имя связывается с ним (x = 3).
Корректные имена: value1, my_var, _temp3.
Некорректное: 5var (имя не может начинаться с цифры).

**2.** Основные типы данных: int (целое число), float (число с плавающей точкой), str (строка), bool (логический тип).

```python
print(type(67))
print(type(2.7))
print(type("What's up?"))
print(type(False))
```

Типизация динамическая: тип есть у значения, а не у переменной; имя в любой момент можно перепривязать к объекту другого типа.

**3.** print() выводит на экран то, что ей передали.

```python
x = 20
print(x)
print("x =", x, "y =", 24)
print(f"Результат: {x + 8}")
```

Разница: через запятую значения разделяются пробелами и типы не нужно приводить; конкатенация требует всё привести к строке; f-строка удобнее всего, когда нам нужен контроль формата.

**4.** Функция type() показывает класс объекта, на который переменная ссылается. Эта функция помогает проверить, с чем реально имеешь дело.

```python
num = 125
name = "Vladimir"
flag = True
print(type(num), type(name), type(flag))
```

```python
print(7+2)    # int
print(7-2)    # int
print(7*2.0)  # float, т.к. один операнд float
print(7/2)    # float, т.к. деление всегда даёт float
print(7//2)   # int для операндов типа int
print(7%2)    # остаток от деления
print(2**10)  # int
```

**6.** Обычное деление всегда даёт float. А целочисленное деление отбрасывает дробную часть, причём для отрицательных чисел - в сторону минус бесконечности (поэтому -7 // 2 = -4, а не -3).

```python
print(7/2, 7//2)
print(-7/2, -7//2)
```

**7.** Ошибка возникает, если строка не является корректной записью числа.

```python
print(int("146"))     # 146
print(float("23.7"))  # 23.7
print(str(67))        # "67"
print(int("abc"))     # ValueError (т.к. нельзя конвертировать нечисловую строку в число)
```

**8.** Любое вычисление можно сразу присвоить или передать в print.

```python
f = 2 + 3 * 4        # Арифметическое выражение (=14)
big = 11 > 3         # Сравнение (True)
ok = big and f > 9   # Логическое выражение (True)
print(f, big, ok)
```

**9.** Приоритет операторов: сначала арифметика, потом сравнения и затем not, and, or

```python
print(4 > 3 and 1 < 5)        # Оба условия истинны
print(not(6 != 5) or 1 < 0)   # False or False => False
print(2+2 == 4 or 5 > 10)     # Сравнение выполнится раньше or
```

**10.** Явное приведение: int + float = float; bool + int = int.
Автоматического приведения нет там, где оно неоднозначно: "5" + 3 вызовет TypeError, потому что надо явно писать int("5") + 3 или "5" + str(3).

**11.** Перечисление через запятую удобно для быстрого вывода; f-строка позволяет контролировать формат; .format() удобно, когда шаблон переиспользуется; конкатенация используется только со строками.

```python
print("Сумма:", 99, "рублей")
print(f"Сумма: {99:.2f} рублей")
print("Сумма: {} рублей".format(99))
print("Сумма: " + str(99) + " рублей")
```

**12.** Приоритет: ** выше, потом унарный минус, затем * / // % и наконец + и -; причём скобки меняют порядок.

```python
print(2 + 3 * 4 ** 2)  # сначала возведение в степень, затем умножение и потом сложение
print((2 + 3) * 4)     # здесь есть скобки: сначала сложение в скобках
```

**14.** Числа, строки, булевы значения и кортежи являются неизменяемыми типами данных, т.е. изменить объект по месту нельзя - любая операция создаёт новый объект.

```python
s = "Привет"
print(id(s))  # один id
s = s + "!"
print(id(s))  # другой id - создана новая строка
```

```python
count = 42              # int - целое число без точки
price = 49.99           # float - есть дробная часть
title = "Товар"         # str - текст в кавычках
total = price * count   # арифметическое выражение, один операнд float, поэтому результат float
print(f"{title}: {total} руб.", type(total))  # вывод через f-строку
print(type(count), type(price), type(title))  # проверка типов через type()
```
